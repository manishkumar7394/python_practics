```python
#Range create a squence of number, which is commonly used in loops and iterative.
# It can take one , two ,or three arguments: START, STOP, STEP.
# the `start` argument is the beginning of the sequence (inclusive)
# the `stop` argument is the end of the sequence (exclusive ),adn
#the `step` argument is the increment between each number in the sequence (default is  1)

ex_range = range(10,50)
print(ex_range)
#note -the range() function is lazy - it genreate a sequence of numbers , but it does not create a list(doesn't sotre all values in memory ).
```

```python
##The range object itself is not a list ,but it can be converted to a list using hte list() function.
list_range =list(ex_range)
print(list_range)

#It is same as using a for loop to iterate through the range object and print each number.

for i in ex_range:
    print(i,end=" ")
    
```

```python

## Length of a range object can be found using the len() function, which returns the number of items in the range.
print("\nLength of ex_range:", len(ex_range))
```

```python
## Range supports slicing, which allows you to create a new range object that is a subset of the original range.
sliced_range = ex_range[5:15]  # This will create a new range
print("Sliced range:", sliced_range)

## Range suppourts negative indexing, which allows you to access elements from the end of the range.
print("Last element of ex_range:", ex_range[-1])  # This will print the last element of the range

## range supports membership testing using the in keyword, which allows you to check if a specific value is present in the range.
print("Is 20 in ex_range?", 20 in ex_range)  # This will return True
print("Is 50 in ex_range?", 50 in ex_range)  # This will return False (because 50 is exclusive)
```

### You can slice list using positive and negative indicies ( mixing them as well) as long as you traverse the list left to right

```python
list_mixed = [10,"ram",59.8,"manish",True,50.3,38,"hari","suraj",20,False,55,"hello" ,59]
print(list_mixed[-6:-2])

print(list_mixed[-2:2])

print(list_mixed[3:-2])  #[first index always start with 0]
```

### [::] displas all the element we can specify step size after that
###### list_mixed[::] will display all the elements list_mixed[::-2] will display all elements skipping by 2 in reverse order

### Three different ways to iterate for loops in python
##### 1. Using in range
##### 2. Using in collections like list, dictionary etc,    (direct member accessed not index)
##### 3. Using enumerate gives access to both index and value


```python
## direct member access using for loop
for i in list_mixed:
    print(i)

## this is internal implementation of list slicing using range object.
## it creates a rang object with the specified start and stop indices, and then iterates through

c_range = range(2,6,1)
for i in c_range:
    print(list_mixed[i],end=" ")

for i in list_mixed:
    print(i)
```

```python
## Enumerate is a built-in function in Python that allows you to loop through a list (or any iterable) and have an automatic counter. 
## It returns both the index and the value of each item in the list during the iteration.

for i , value in enumerate(list_mixed):
    print(f"Index: {i}, Value:  {value}")
```

```python
## Because loops have in built if-else construct, we can use else with for loop. 
## The else block will be executed after the for loop completes normally (i.e., without encountering a break statement).

for num in range(1,4):
    if num>1:
        for i in range(2,num):
            if num%i==0:
                break
        else:
            print(num)
```
