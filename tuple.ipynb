# 2.TUPLE- Order ,Immutable Collection
##### A  `Tuple   ` is like a list that __cannot be changed__ after creation. This makes it:
- **Safer for data that should not be modified**
- **Usable as dictionary keys (list cannot be used as keys)**
- **Slightly faster than lists for iteration**

```python
## Creating Tuples
coordinates = (10.0, 20.0)
rgb_color =(255,0,0)
single= (42,) # Single element tuple requires a comma
from_list = tuple([1,2,4,5]) # Convert list to tuple

print(f"Coordinates: {coordinates}")
print(f"RGB Color: {rgb_color}")
print(f"Single Element: {single}, Type: {type(single)}")
print(f"From List: {from_list}, Type: {type(from_list)}")
```

```python
## Accessing and Unpacking Tuples
point = (10, 20, 30)

#Accessing by index
print(f"first : {point[0]}")
print(f"last : {point[-1]}")

# Unpacking   -  assigning tuple elements to variables
x, y, z = point
print(f"x: {x}, y: {y}, z: {z}")

#Unpacking with * (star) operator
first, *rest =(1,24,5,6,7)
print(f"first: {first}, rest: {rest}")

##tuple are immutable - this would raise type error
#point[0] = 100  # Uncommenting this line will raise a TypeError
```

```python
#Tuple methods and operations
data = (1, 2, 3, 2, 4, 2)

print(f"count of 2: {data.count(2)}")
print(f"index of 3: {data.index(3)}")
print(f"Length of tuple: {len(data)}")
print(f"Concatenation: {data + (5, 6)}")    #adding tuples
print(f"Repetition: {data * 2}")             #repeating tuples
print(f"Membership: {2 in data}, {10 in data}")  #checking membership

#common use: returning multiple values from a function
def min_max(numbers):
    return min(numbers), max(numbers)   #returning a tuple
    lo, hi = min_max([3, 1, 4, 1, 5, 9])
    
    print(f"Min: {lo}, Max: {hi}")           
```
