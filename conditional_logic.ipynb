# Conditional Logic in Python
##### Conditional statements let your program make __decisions__ execute different blocks of code based on whether a conditions is `True`or`False`

## Tools Covered in This Notebook
1. `if` - single condition
2. `if...else` - two branches
3. `elif` - multiple branches 
4. `match..case` -structural pattern matching 
5. Ternary operator- one-line conditional expression


## 1. `If` - Execute Only when True
##### The simplest conditional. the indented block runs only if the condition evaluates to `True`.

```python
## I - basic condition check
temperature = 38
if temperature >37:
    print("you have a fever.pls rest!")
#Nothing prints if temperature<=-the block is simply skipped 
age = 20
if age >= 18:
    print("You are eligible to vote.")
```

## 2.if...else - Two Branches
#### When you need to do one thing if `True` and another if `False`

```python
# If....two possible outcomes
number =15
if number %2 ==0:
    print(f"{number} is even")
else:
    print(f"{number} is odd")

#Chacking loging status
is_logging_in = False

if is_logging_in:
    print("welcome back")
else:
    print("pls log in to countinue.")
```

## 3.`elif`- Muliple Branches
##### When there are more than two possiblities, use `elif` (short for "else if " ) Python checks each condition from top to bottom and exutes the first one that is   `True`.

```python
#elif - grading system 
score = 30
if score >=90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >=70:
    grade = "c"
elif score >=60:
    grade = "D"
else:
    grade = "f"

print(f"Score :{score}  -> Grade:{grade}")
```

```python
## Cobining conditions with 'and', 'or', 'not'ArithmeticError

age = 18
has_license = True

if age >=18 and has_license:
    print("You can drive")
age = 45
if age <13 or age >65:
    print("you get a discount.")
else:
    print("standerd pricing applies")

is_blocked = False
if not is_blocked:
    print("Access granted.")
```

## 4. `Match..case` - Structure pattern Matching
##### `match` is Python's answer to switch in other languages, but far more powerful. It matches patterns, not just values

```python
#match....case - basic value matching

command = "start"

match command:
    case "start":
        print("Starting the application...")
    case "stop":
        print("Stopping the application...")
    case _:
        print(f"Unknown command:{command}")
```

```python
#match..case - Matching with patterns and guards
#Matching HTTP status codes

status_code = 404

match status_code:
    case 200:
        print("OK")
    case 301:
        print("Redirect")
    case 404:
        print("Not Found")
    case 500:
        print("Internal Server Error")
    case code if 400 <= code <500:  #guard clause -additional condition
        print(f"Client Error")
    case _:
        print(f"Other status: {status_code}")

```

```python
##match...case - Matching data structures(the real power)

point = (0, 5)

match point:
    case (0,0):
        print("origin")
    case (x, 0):
        print(f"on the x-axis at x= {x}")
    case (0, y):
        print(f"on the y-axis at y={y}")
    case (x, y):
        print(f"point at {x},{y}")
```

## 5. Ternary Operator -One line conditional
 `yntax: value_if_true if condition else value_if_false`

##### Use this for simple assignments or expressions. For complex logic, stick with full if/else blocks — readability matters

```python
## Ternary Operator - concise conditional expressions

age = 34
status = "adult" if age >=18 else "minor"
print(f"Age {age} - {status}")

## practical uses
score = 85
result = "pass" if score >= 60 else "fail"
print(f"score {score} - {result}")

## Inside an f-string
count = 1
print(f"you have {count} {'item' if count ==1 else "item"} in your cart")

# Nested ternary (use sparingly - can hurt readability)
x= 0
sign = "positive" if x > 0 else "nagetive" if x<0 else "zero"
print(f"{x} is {sign}")


```
