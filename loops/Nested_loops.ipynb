# NESTED LOOPS 
#### A loop inside a loop . The innder loop run completely for each iteration of the outer loop.


```python
#Nested loops - multiplication table

for i in range(1,11):
    for j in range(1,11):
        print(f"{i*j:4}", end="")
    print()         ##newline after each row

print("----")
```

```python
# Practical example: finding pairs that sum to a target
numbers = [1, 3, 5, 7, 9]
target_sum = 10

for i in range(len(numbers)):
    for j in range(i + 1, len(numbers)):   # Start from i+1 to avoid duplicates
        if numbers[i] + numbers[j] == target_sum:
            print(f"{numbers[i]} + {numbers[j]} = {target_sum}")
```
