# 1. `for` with  `range()` - Repeat a Fixed Number of Times

#### `range (start, stop, step)` generates a sequence of numbers.
- `start` defaults to `0`
- `stop` is exclusive(not included)
- `step` defaults to `1`


```python
## range() - basic usage

# Print 0 through 4
for i in range(5):
    print(i, end=" ")       #[end=" "] - space in element
print() #newline

# range(start, stop) - from 2 to 6
for i in range(2,7):
    print(i, end=" ")
print() 

#range(start, step, stop) - count by 3s
for i in range(0,20,3):
    print(i, end=" ")
print()

# Counting backwards 
for i in range(10, 0, -2):
    print(i, end=" ")
print()
```

## 2. `for` Over a List
##### the most common use of `for` - iterate over every element in a list(or any iterable).


```python
## Iterating over  a List

fruits = ["apple","banana","cherry","date"]

for fruit in fruits:
    print(f"I like {fruit}")

print("-------")

# Iterating over a string (string are iterable too!)
for char in "Python" :
    print(char, end="_")
print()

## Summing a list of number
prices = [10.5, 20.0, 35.75, 8.25]
total = 0
for price in prices:
    total += price
print(f"Total : ${total:.2f}")          # {:.2f} = add decimal number

# Tips : Python's built -in sum() does this in one call
print(f"Total (built-in) : ${sum(prices):.2f}")


```

## 3.`enumerate()` - Get Index and value Together
##### Instead of using `range(len(list))` to track the index, use `enumerate`. it is more Pythonic and less error- prone


```python
## enumerate() - the Pythonic way to get index + value
languages = ["Python","JavaScript","Rust","Go"]

# The old way (avoid this)
print("--- Using range(len()) ---")
for i in range(len(languages)):
    print(f"{i}: {languages[i]}")

# The Pythonic way
print(f"--- Using enumerate() ---")
for index, lang in enumerate(languages):
    print(f"{index}: {lang}")

# Start counting form 1 instead of 0
print("--- Starting from 1 ---")
for rank, lang in enumerate(languages,start=1):
    print(f"#{rank}: {lang}")
```

## 4. `for` Over a Dictionary
##### Dictionary offer multiple ways to iterate: Over keys, values, or key- value pairs.



```python
## Iterating Over a dictionary 

scores = {"Alice": 90, "bob": 85, "Charlie": 93, "Diana": 53}

# Iterate over keys (default)
print(f"--- keys ----")
for name in scores:
    print(f"name: {name}")

# Iterate Over values
print("--- Values ---")
for score in scores.values():
    print(f"score: {score}")

# Iterative over key - value pairs (most useful)
print("--- Item ---")

for name, score in scores.items():
    status = "Pass" if score >= 60 else "Fail"
    print(f"{name}: {score} ({status})")

# find the top scorer
top_name = max(scores, key=scores.get)
print(f"\nTop scorer: {top_name} With {scores[top_name]}")


# find the low scorer
low_name = min(scores, key=scores.get)
print(f"\nLower Scorer: {low_name} with {scores[low_name]}")

```
