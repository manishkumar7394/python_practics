# 5. `while` Loop - Repeat Until a Condition Changes
##### A While loop keeps running as long as its condition is `True`. Be careful - if the condition never becomes `False` , you get an __infinite loop__

```python
## While loop - basic countdown

count = 5
while count > 0:
    print(f"countdown : {count}")
    count -= 1
print("Liftoff!")

print("-----")

# Practical example: keep doubling untill we exceed 1000
value = 1
step = 0 
while value <= 1000:
    value *= 2
    step += 1
print(f"Reached {value} in {step} doublings")
```

```python
var_condition = input("Enter a value T/F :")
while var_condition != "F":
    print (f"value  is still {var_condition}")
    var_condition = input("Enter a value T/F, enter F to exit")
else:
    print("Uster entered F exiting now")
```
