# LOOPS IN PYTHON  - For and While

1. `For` with `range()`

4. `for ` over a Dictionary
5. `while` loops
6. `break`, `continue` and `else ` on loops
7.  Nested loops
8. `zip()` - iterate over multiple sequences in parallel
