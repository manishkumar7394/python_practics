# SETS
### Sets are in- build data type in Python - to store collections of Unique items
### Unordered (items do not follow any arrangement of ascendiing or descending)
### Do not allows duplicates elements
### usesd for  membership tests, eliminating duplicates values, performing set operations like union, intersection difference and symmetric defference


```python
# creating sets using {}, just like dictionary its not key value pair 
set1 = {1,2,3,4,5,8}
print(set1) 
print(type(set1))
```

```python
# creating empty sets
set2 ={} #this will create empty dictionary -- be carefull
set3 =set()
print(set2,set3)
print(type(set2),type(set3))
```

```python
# Basic set operations
# Adding new elements using add method
set3.add(3)
set3.add(4)
set3.add(6)
set3.add(5)
set3.add(9)
set3.add(7)
set3.add(11)
set3.add(23)
set3.add(1)
set3.add(4)         #will not add as its already there

print(f"add all element in one set: {set3}")
```

```python
## Removing element from set
set3.discard(9)         #remove 9 element form set3
set3 
```

```python
set3.pop() # pop is first in first out fifo
set3
```

```python
# Membership test - using in operator
print(10 in set3)
print(15 in set3)
```

```python
#set operations - Union , intersection, difference, sysmentic difference, subset,superset
set1 = {1,2,3,4,5,6,7,8}
set2 = {4,3,5,8,9,4,10}
union_set = set1.union(set2) #all the unique element form a and b set
print(union_set)

intersect_set= set1.intersection(set2)  #takes the common elmnt from a and b
print(intersect_set)

difference_set= set1.difference(set2)     #keep a and remove common element from b
print(difference_set)

difference_set= set2.difference(set1)       #keep a -b is not same as b-a
print(difference_set)

result =set1.symmetric_difference(set2)         #keep union - the intersection results.
print(result)

result= set2.symmetric_difference(set1)
print(result)
```

```python
set1.intersection_update(set2) #update in place set operations
print(set1)

### subset return always boolian value
print(set1.issubset(set2))    
print(set2.issubset(set1))
```
