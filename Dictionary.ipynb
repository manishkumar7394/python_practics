# Dictionary - Key-value Pairs
#### A `dict   ` store data as **key:value** pairs. Keys must be unique and immutable (strings , numbers, tuples).Values can be anything.


```python
##Creating Dictionaries
student = {
    "name":     "Manish Paswan",
    "age":      "23",
    "course":   "AI Internship",
    "Active" :  True
}


print(f"student: {student}")
print(f"Name: {student['name']}")
print(f"keys:{list(student.keys())}")
print(f"values:{list(student.values())}")
print(f"items :{list(student.items())}")
```

```python
## Accessing values- [] vs.get()
student = {
    "name":    "Manish Paswan",
    "age":      23
}
#direct access using - raise key error if key is missing
print(student['name'])

## Safe access using get() - returns None if key is missing
print(student.get('email'))  # returns None
print(student.get('email','not found'))  # returns 'Not Found' if key is missing   

### Tip: Use .get() when you are not sure if the key exists in the dictionary to avoid KeyError.
```

```python
# modifying Dictionaries
student = {
    "name":     "Manish Paswan",
    "age":      23
}
# Add a new key-value pair
student['course'] = 'AI Internship'
student['age'] = 24  # Update existing key
print(f" Updated:    {student}")

# Remove
del student["course"]  # delete by key
removed = student.pop("age")    #remove and return value
print(f"Removed: {removed}, Updated: {student}")


# Marge two dictionaries
defaults = {'theme': 'light', 'language': 'english'}
overrides = {'theme': 'dark','font_size': 12}
marged = {**defaults, **overrides} # Merge dictionaries
print(f"Merged: {marged}")
```

```python
#Iterative Over a Dictionary
scores = {'Alice': 90, 'Bob': 85, 'Charlie': 92}
# Iterate over keys (default)
for name in scores:
    print(f"{name}: {scores[name]}")

print("-----")    

## Iterate over key -vlaue pairs (preferred)
for name, score in scores.items():
    print(f"{name}: {score}")
print("------")   

### Dictionary Comprehension
doubled ={name: score * 2 for name, score in scores.items() }
print(f"Doubled: {doubled}")
```

## QUICK COMPARISION :  when to use what?

| Need | Use |
|---|---|
| Ordered collection that changes | **List** |
| Fixed data, function return values | **Tuple** |
| Fast lookup by key | **Dictionary** |
| Unique items only | **Set** `{ }` (bonus — covered in a later lesson) 

**Key Insight:** Lists and Dicts are passed by **reference** in Python. If you pass a list to a function and modify it inside, the original list is also modified. Use `.copy()` or `list()` to create a shallow copy when you need independence.
