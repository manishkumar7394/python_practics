```python
def prime_test(divident):
    if divident > 1 :
        for test_all_nums in range (2, divident):
            if divident % test_all_nums == 0 :
                return 0
                break 
            else: 
                return 1
        else:
            return 0
prime_numbers = [x for x in range(100) if prime_test(x)==1]

prime_numbers
```

```python
print(f"Prime test for number {10} is {prime_test(10)}")
```

```python
def prime_test_print(x):
    if prime_test (x):
        print(f"Number {x} is a prime number")
    else:
        print(f"number {x} is not a prime number")

prime_test_print(29)
prime_test_print(30)
```
