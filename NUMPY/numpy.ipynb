```python

import time

start_time = time.perf_counter()

rows_a = len(list1)
cols_a = len(list1[0])
rows_b = len(list2)
cols_b = len(list2[0])

if cols_a != rows_b:
    print ("Invalid matrix dimensons")
else:
    list3 = [[0 for _ in range(cols_b)] for _ in range(rows_a)]
    for row in range(rows_a):
        for col in range(cols_b):
            sum_product =0
            for element in range(cols_a):
                sum_product += list1[row][element] * list2[element][col]
            list3[row][col] =sum_product
    print (list3)

tot_run_time = time.perf_counter() - start_time
print(f"Total elapsed time is : {tot_run_time}")
```
