```python
class shape:
    #len, width = 1,10
    # def - 
    def __init__(self, len, width):
        self. len = len
        self.width = width
    
    def area(self):
        return self.len* self.width
    
    def paint_cost(self, cost_perunit):
        return self.area() * cost_perunit
```

```python
s = shape(3,5)
print(f"Area of shape is {s.area()}")
print(f"cost of painiting this shape is {s.paint_cost(3)}")
```

```python
class dog:
    def __init__(self, name, age):
        self.name = name
        self. age= age
    
    def massage(self):
        print("I am a dog")
        print(f"My name is {self.name}")
        print(f"My age is {self.age}")
    
    def sleep(self):
        print("sleeping")

    def speak(self):
        print("Bark")
```

```python
class pet:
    def __init__ (self, name, age, gender):
        self.name= name
        self.age = age
        self.gender= gender
    
    def massage(self):
        print(f"My name is {self.name}")
        print(f"My age is {self.age} And my gender is {self.gender}")

    def sleep(self):
        print("not more sleeping")

    def wakeup(self):
        self.sleep()
        print("Wakeup")
    
    
class dog(pet):
    def __init__ (self, name, age, gender,breed):
        super().__init__(name, age, gender)
        self. breed = breed
   
    def speak(self):
        print(f"I am a dog of breed {self.breed}")
        print("I bark")

class cat(pet):
    def __init__(self, name , age , gender, breed):
        super().__init__(name, age, gender)
        self.breed = breed
    
    def speak(self):
        print(f"I am a Cat and my breed {self.breed}")

        print("I Meaow")
```
