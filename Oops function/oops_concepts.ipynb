### Reference Variavles
- Reference variavble hold the objects
- We can creare objects without reference variable as well
- an object can have multiple reference variable
- Assigning a new reference variable to an existing object object does not create a new object


```python
# Object without a reference 
class person:
    def __init__ (self):
        self.name = 'manish'
        self.gender  = 'male'
    
    
```

```python
## multiple refenece
p = person()
# or
p.name   # reference of perosn
q = p     # multi reference 
q.name

```

```python
## change attribute value with the help of 2nd object
print(p.name)
print(q.name)
q.name = 'manish paswan'
print(q.name)
print(p.name)
```

```python
class Person:
    def __init__(self,name, gender):
        self.name = name
        self.gender = gender

# outside the class -> function
def greet(person):
    print('Hi my name is', person.name, 'and I am a ',person.gender)
    p1 = Person('ankita','female')
    return p1
    
p = person('manish','male')
x = greet(p)
print(x.name)
print(x.gender)
```
