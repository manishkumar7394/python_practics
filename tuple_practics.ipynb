```python
#list can be coverted into tuples and tuples can be converted back to list
list =[1,3,4,5,6]
print(type(list))
tuple_eg = tuple(list)
print(tuple_eg)
print(type(tuple_eg))
```

```python
print(tuple_eg[::])
#similar slicing operations as list, just return type is tuple
```

```python
#Tuple operations
#concatenation using +operation
mixed_t = ("hello","world")
concat_tuple= mixed_t + tuple_eg
concat_tuple
```

```python
mult_tuple = mixed_t*3
mult_tuple
```

```python
#immutabel nature of tuple 
tuple_eg[2]= 10 # will give an error
```
